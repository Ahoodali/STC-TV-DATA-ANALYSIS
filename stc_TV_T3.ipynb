```python
"""
TODO show the recommendations (top 5) for the people who watched "Moana" movie
"""
```

```python
moana_recommendations = dataframe[
    dataframe["user_id_maped"].isin(
        dataframe.loc[
            dataframe["program_name"] == "Moana",
            "user_id_maped"
        ]
    )
]

moana_recommendations = (
    moana_recommendations[
        moana_recommendations["program_name"] != "Moana"
    ]
    .groupby("program_name")["rating"]
    .mean()
    .sort_values(ascending=False)
    .head(5)
)

moana_recommendations
```

```python
recommendation_sample = moana_recommendations.reset_index()
recommendation_sample.columns = ["Recommended_Program", "Average_Rating"]

recommendation_sample
```
