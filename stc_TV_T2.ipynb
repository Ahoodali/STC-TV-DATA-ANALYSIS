```python
"""
TODO using the previous dataset (df) build a prediction model to predict the expected watch time for the next two months
Hint: you can build a forecast model to predict the results
"""
```

```python
df=pd.read_excel("stc- Jawwy TV Data Set_T2.xlsb")
#
```

```python
df.columns
```

```python
df["date_"]=pd.to_datetime(df["date_"])
```

```python
monthly_watch = df.groupby(df["date_"].dt.to_period("M"))["Total_watch_time_in_houres"].sum()

monthly_watch
```

```python
monthly_watch.index = monthly_watch.index.to_timestamp()

monthly_watch
```

```python
len(monthly_watch)
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing

model = ExponentialSmoothing(monthly_watch, trend="add")
fit = model.fit()

forecast = fit.forecast(2)

forecast
```

```python
forecast_df = forecast.reset_index()
forecast_df.columns = ["Date", "Predicted_Watch_Time_in_Hours"]

forecast_df
```

```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=monthly_watch.index,
    y=monthly_watch.values,
    mode="lines+markers",
    name="Actual Watch Time"
))

fig.add_trace(go.Scatter(
    x=forecast.index,
    y=forecast.values,
    mode="lines+markers",
    name="Forecast"
))

fig.update_layout(
    title="Watch Time Forecast for the Next Two Months",
    xaxis_title="Date",
    yaxis_title="Watch Time in Hours"
)

fig.show()
```

Finding:

A forecasting model was built using the historical monthly watch time data.
The model predicts the expected total watch time for the next two months based on the previous viewing pattern.

The forecast results are shown in the table and visualization above.
